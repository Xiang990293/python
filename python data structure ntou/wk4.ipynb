```python
x = ["I", "II", "III", "IV", "V", "VI"]
y = [21,8,8,4,4,4]

plt.figure(figsize=(8,4))

bars = plt.bar(x, y, color=["blue", "green", "#FF7700", "purple", "#FF77FF", "#aaaaaa"], edgecolor="black", linewidth=1, width=.5)

for bar in bars:
    height = bar.get_height()
plt.title("Typhoon track class")
plt.ylabel("Number")
plt.yticks([0, 5, 10, 15, 20])
plt.bar_label(bars, labels=y)
```
