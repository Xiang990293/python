```python
import numpy as np
```

```python
one = np.array(range(20, 20+9))

print(*one[0:6])
```

```python
two = np.array([[20,21,22],
                [23,24,25],
                [26,27,28]])
print(*[int(two[i][i]) for i in range(len(two))])
```

```python
print(np.arange(20,20+10))
print(np.arange(20,20+10,2))
print(np.arange(20,20+10,3))
```

```python
print(np.zeros_like(two))
print(np.ones_like(one))
```

```python
two = np.arange(20,20+36).reshape(9,4)
print(two, "\n")
two = np.arange(20,20+36).reshape(4,9)
print(two, "\n")
two = np.arange(20,20+36).reshape(3,12)
print(two, "\n")
two = np.arange(20,20+36).reshape(12,3)
print(two, "\n")
```

```python
ten = np.arange(20,20+10).reshape(2,5)
print(f"{type(ten)= }, \n{ten.dtype= }, \n{ten.size= }, \n{ten.shape= }, \n{ten.itemsize= }, \n{ten.ndim= }, \n{ten.nbytes= }")
```
